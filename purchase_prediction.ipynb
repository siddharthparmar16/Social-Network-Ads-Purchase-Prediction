# Social Network Ads Purchase Prediction

## Goal
The goal of this project is to predict whether a user will purchase a product based on their gender, age, and estimated salary.

## Dataset
The dataset contains the following columns:
- Gender
- Age
- EstimatedSalary
- Purchased

## Model Used
I used Logistic Regression because this is a binary classification problem.

## Result
The model achieved around 88% accuracy on the test data.

```python
import pandas as pd
import numpy as np
```

```python
from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, MinMaxScaler
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score
```

```python
df = pd.read_csv('Social_Network_Ads.csv')
```

```python
df
```

```python
df.drop(columns=['User ID'], inplace=True)
```

```python
df
```

```python
df.isna().sum()
```

```python
X_train,X_test,y_train,y_test = train_test_split(df.drop(columns=['Purchased']),
                                                 df['Purchased'],
                                                 test_size = 0.2,
                                                 random_state = 42,
                                                 stratify=df['Purchased'])
```

```python
X_test
```

```python
trf1 = ColumnTransformer([
    ('ohe_gender',OneHotEncoder(sparse_output=False,handle_unknown='ignore'),[0])
],remainder='passthrough')
```

```python
trf1
```

```python
df
```

```python
trf2 = ColumnTransformer([
    ('scale',MinMaxScaler(),slice(0,3))
])
```

```python
trf3 = LogisticRegression()
```

```python
pipe = Pipeline([
    ('trf1',trf1),
    ('trf2',trf2),
    ('trf3',trf3)
])
```

```python
# train
pipe.fit(X_train,y_train)
```

```python
pipe.named_steps
```

```python
y_pred = pipe.predict(X_test)
```

```python
y_pred
```

```python
accuracy_score(y_test,y_pred)
```

```python
from sklearn.metrics import confusion_matrix, classification_report

print(confusion_matrix(y_test, y_pred))
print(classification_report(y_test, y_pred))
```

## Conclusion

The Logistic Regression model achieved an accuracy of 88% on the test data.  
The model performed well in predicting users who did not purchase, with a recall of 94% for class 0.  
For users who purchased, the recall was 76%, meaning the model correctly identified most buyers but missed some.  

Overall, this is a good baseline classification model for predicting whether a user will purchase based on gender, age, and estimated salary.
